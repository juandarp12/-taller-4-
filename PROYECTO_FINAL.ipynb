```python
Ventas=[1200,1500,1752,1425,2001,1015]

def calcular_promedio(lista):
  total=sum(lista)
  cantidad=len(lista)
  return total/cantidad

def clasificar_ventas(promedio):
  if promedio > 1500:
    return "Alto"
  elif promedio >= 1050:
    return "Medio"
  else:
    return "Bajo"


promedio_ventas=calcular_promedio(Ventas)
clasificacion=clasificar_ventas(promedio_ventas)
print("Nivel de ventas:", clasificacion)
```

```python
import matplotlib.pyplot as plt

# Create a bar chart for sales data
plt.figure(figsize=(10, 6))
bars = plt.bar(range(len(Ventas)), Ventas, color='lightcoral')
plt.xlabel('Venta Individual')
plt.ylabel('Monto de Venta')
plt.title('Monto de Ventas Individuales')
plt.xticks(range(len(Ventas)), [f'Venta {i+1}' for i in range(len(Ventas))], rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on top of the bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 50, round(yval, 2), ha='center', va='bottom') # Adjust +50 for label height

plt.tight_layout()
plt.show()
```
