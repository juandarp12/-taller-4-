```python
ventas = [200, 350, 150, 400, 300]

total_ventas = sum(ventas)
promedio = total_ventas / len(ventas)
mayores_300 = len([v for v in ventas if v > 300])

if promedio > 300:
    clasificacion = "Alto"
    conclusion = "Está bien"
elif promedio >= 200:
    clasificacion = "Medio"
    conclusion = "Hay riesgo"
else:
    clasificacion = "Bajo"
    conclusion = "Debe mejorar"

print(f"Total: {total_ventas}")
print(f"Promedio: {promedio}")
print(f"Días > 300: {mayores_300}")
print(f"Clasificación: {clasificacion}")
print(f"Conclusión: {conclusion}")
```
